Merge the datasets

```python
claims_df = delay_register.merge(
    performance_df,
    on=["task_id", "task_code", "task_name"],
    how="left"
)

print(claims_df.shape)
claims_df.head()
```

Potential EOT

```python
claims_df["potential_eot"] = np.where(
    claims_df["delay_type"] == "Employer",
    "Yes",
    np.where(
        claims_df["delay_type"] == "Exceptional Event",
        "Potential",
        np.where(
            claims_df["delay_type"] == "Contractor",
            "No",
            "Requires Review"
        )
    )
)
```

```python
claims_df[
    [
        "delay_id",
        "task_code",
        "task_name",
        "delay_event",
        "delay_days",
        "delay_type",
        "potential_eot",
        "notice_issued",
        "evidence_available"
    ]
].head(20)
```

```python
# ============================
# Executive KPIs
# ============================

total_delay_events = len(claims_df)

potential_eot_cases = (
    claims_df["potential_eot"] == "Yes"
).sum()

potential_exceptional_cases = (
    claims_df["potential_eot"] == "Potential"
).sum()

total_delay_days = claims_df["delay_days"].sum()

executive_kpis = pd.DataFrame({
    "KPI": [
        "Total Delay Events",
        "Total Delay Days",
        "Potential EOT Cases",
        "Potential Exceptional Event Cases"
    ],
    "Value": [
        total_delay_events,
        total_delay_days,
        potential_eot_cases,
        potential_exceptional_cases
    ]
})

executive_kpis
```

```python
# ============================
# Export Claims & EOT Report
# ============================

claims_report = claims_df[
    [
        "delay_id",
        "task_code",
        "task_name",
        "delay_event",
        "delay_days",
        "delay_type",
        "potential_eot",
        "notice_issued",
        "evidence_available"
    ]
].copy()

with pd.ExcelWriter(
    "FIDIC_Claims_EOT_Intelligence_Report_V01.xlsx"
) as writer:

    executive_kpis.to_excel(
        writer,
        sheet_name="Executive_KPIs",
        index=False
    )

    claims_report.to_excel(
        writer,
        sheet_name="Claims_Intelligence",
        index=False
    )

print("FIDIC Claims & EOT report exported successfully.")
```
